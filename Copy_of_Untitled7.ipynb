#method 2
2) Intrinsic NumPy array creation functions

```python
# 1d
# making it normally
m2=np.arange(10)
print(m2)
print("\n")
print("next methood ")
print("\n")
# method where we define start stop and data type as well
a= np.arange(10,78,dtype= np.float128 )
print(a)
print("\n")
print("next methood \n")
# next method where we deine start stop and step values as well
aa= np.arange(10,78,2,dtype= np.float128 )
print(aa)
```

```python
"""When we use np.arange(start, stop, step), the stop value is generally not included.
The number of elements also depends on the step size. So, if we want to specify exactly how many values we
 need while also including both the starting and ending values, we can use np.linspace(start, stop, num)"""
#  line space
bb= np.linspace(10,78,10)
print(bb)
```
